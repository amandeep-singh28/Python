```python
# Increasing Triangle
n = 5
for i in range(n):
    for j in range(0, i + 1):
        print('*', end = ' ')
    print()
```

```python
# Decreasing Triangle
n = 5
for i in range(n):
    for j in range(i, n):
        print('*', end = ' ')
    print()
```

```python
# Right Sided Triangle
n = 5
for i in range(n):
    for j in range(i, n):
        print(' ', end = ' ')
    for j in range(0, i + 1):
        print('*', end = ' ')
    print()
```

```python
n = 5
for i in range(n):
    for j in range(0, i + 1):
        print(' ', end = ' ')
    for j in range(i, n):
        print('*', end = ' ')
    print()
```

```python
# Hill pattern
n = 5
for i in range(n):
    for j in range(i, n):
        print(' ', end = ' ')
    for j in range(0, i + 1):
        print('*', end = ' ')
    for j in range(1, i + 1):
        print('*', end = ' ')
    print()
```

```python
# Reverse Hill pattern
n = 5
for i in range(n):
    for j in range(0, i + 1):
        print(' ', end = ' ')
    for j in range(i, n):
        print('*', end = ' ')
    for j in range(i, n - 1):
        print('*', end = ' ')
    print()
```

```python
# Diamond pattern
n = 5
for i in range(n - 1):
    for j in range(i, n):
        print(' ', end = ' ')
    for j in range(0, i + 1):
        print('*', end = ' ')
    for j in range(1, i + 1):
        print('*', end = ' ')
    print()
for i in range(n):
    for j in range(0, i + 1):
        print(' ', end = ' ')
    for j in range(i, n):
        print('*', end = ' ')
    for j in range(i, n - 1):
        print('*', end = ' ')
    print()
    
```
